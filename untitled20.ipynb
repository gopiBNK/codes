```python
from sqlalchemy import create_engine
import pandas as pd
import zipfile
import json
import os
import shutil

# MySQL database connection details
db_user = 'root'
db_password = 'gopinathv.p02@'
db_host = 'localhost'  # or your MySQL server address
db_name = 'cricket'

# Path to the ZIP file
zip_file_path = 't20s_json.zip'

# List to store extracted delivery data
delivery_data = []

# Open the ZIP file
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    # Extract all files to a temporary directory
    temp_dir = 'temp_extracted'
    zip_ref.extractall(temp_dir)

    # Loop through each file in the extracted directory
    for match_id, file_name in enumerate(os.listdir(temp_dir), start=1):  # Add a unique match_id
        if file_name.endswith('.json'):  # Process only JSON files
            file_path = os.path.join(temp_dir, file_name)

            # Read the JSON file
            with open(file_path, 'r') as json_file:
                json_data = json.load(json_file)

                # Extract innings data
                for inning in json_data.get("innings", []):
                    team = inning.get("team", None)
                    for over_info in inning.get("overs", []):
                        over_number = over_info.get("over", None)
                        for delivery in over_info.get("deliveries", []):
                            runs = delivery.get("runs", {})
                            delivery_record = {
                                "team": team,
                                "over": over_number,
                                "batter": delivery.get("batter", None),
                                "bowler": delivery.get("bowler", None),
                                "non_striker": delivery.get("non_striker", None),
                                "batter_runs": runs.get("batter", 0),
                                "extras_runs": runs.get("extras", 0),
                                "total_runs": runs.get("total", 0),
                            }
                            delivery_data.append(delivery_record)

# Convert the extracted data into a DataFrame
delivery_df = pd.DataFrame(delivery_data)

# Clean up temporary directory
shutil.rmtree(temp_dir)

# Establish a connection to the MySQL database
engine = create_engine(f'mysql+mysqlconnector://{db_user}:{db_password}@{db_host}/{db_name}')

# Create the table and insert data
table_name = 't20_deliveries'
delivery_df.to_sql(name=table_name, con=engine, if_exists='replace', index=False)

print(f"Data inserted successfully into table '{table_name}' in the database '{db_name}'.")
```
